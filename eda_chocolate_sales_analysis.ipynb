```python
import numpy as np
import pandas as pd
#import matplotlib.plyplot as plt
import seaborn as sns

data = pd.read_csv("Cleaned_Chocolate_Sales.csv")

print(data.head())
print(data.shape)
print(data.info())
print(data.describe())
print(data.columns)

print("Missing values:")
print(data.isnull().sum())

print("Duplicate rows:", data.duplicated().sum())

print("Total Sales:", data["Amount"].sum())
print("Average Sales:", data["Amount"].mean())
print("Highest Sale:", data["Amount"].max())
print("Lowest Sale:", data["Amount"].min())

print("Total Boxes Shipped:", data["Boxes Shipped"].sum())
print("Average Boxes Shipped:", data["Boxes Shipped"].mean())

country_sales = data.groupby("Country")["Amount"].sum().sort_values()
print(country_sales)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Chocolate_Sales.csv")

# Calculate total sales by product
product_sales = data.groupby("Product")["Amount"].sum().sort_values()

# Create bar chart
plt.figure(figsize=(12, 6))
plt.bar(product_sales.index, product_sales.values,width=0.6, color="chocolate")

plt.title("Total Chocolate Sales by Flavour")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Chocolate_Sales.csv")
data["Date"] = pd.to_datetime(data["Date"])

data["Month"] = data["Date"].dt.to_period("M")

monthly_sales = data.groupby("Month")["Amount"].sum()

print(monthly_sales)

monthly_sales.plot(kind="line", marker="o", figsize=(12, 4))

plt.title("Monthly Chocolate Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Chocolate_Sales.csv")
boxes_by_product = data.groupby("Product")["Boxes Shipped"].sum().sort_values()
print(boxes_by_product)

boxes_by_product.plot(kind="bar", color="chocolate",figsize=(12, 5))

plt.title("Total Boxes Shipped by Product")
plt.xlabel("Product")
plt.ylabel("Boxes Shipped")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Chocolate_Sales.csv")
print(data[["Amount", "Boxes Shipped"]].corr())

plt.scatter(data["Boxes Shipped"], data["Amount"])

plt.title("Sales Amount vs Boxes Shipped")
plt.xlabel("Boxes Shipped")
plt.ylabel("Sales Amount")
plt.tight_layout()
plt.show()
```
